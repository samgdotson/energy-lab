```python
import atlite
```

```python
minx, miny, maxx, maxy = [-92.889433,  42.49172 , -86.249548,  47.309822]
```

```python
TIME = ('2015-01-01','2015-12-31')
cutout = atlite.Cutout(
    path='wisconsin-2015.nc',
    module='era5',
    x=slice(minx, maxx),
    y=slice(miny, maxy),
    time=slice(*TIME)
)
```

```python
cutout.prepare()
```
